# Movie Data Analysis

```python
import pandas as pd
import matplotlib
matplotlib.use('module://matplotlib_inline.backend_inline')
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv('movies.csv')

print("Dataset shape:", df.shape)
df.head()
```

## 1. Column Charts


```python
top_movies = df.nlargest(10, 'Box Office (M$)')
plt.figure(figsize=(12, 6))
plt.barh(top_movies['Title'], top_movies['Box Office (M$)'], color='steelblue')
plt.xlabel('Box Office Revenue (Million $)')
plt.ylabel('Movie Title')
plt.title('Top 10 Movies by Box Office Revenue')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
plt.close()
```

```python
genre_box_office = df.groupby('Genre')['Box Office (M$)'].mean().sort_values(ascending=False)
plt.figure(figsize=(10, 6))
plt.bar(genre_box_office.index, genre_box_office.values, color='coral')
plt.xlabel('Genre')
plt.ylabel('Average Box Office Revenue (Million $)')
plt.title('Average Box Office Revenue by Genre')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
plt.close()
```

### Interpretation of Column Charts

**Top 10 Movies by Box Office Revenue**: Avatar: The Way of Water leads with 2.32 billion, followed by Spider Man No Way Home at $1.92 billion. The chart shows that action and scifi films dominate the top positions, with most top earners exceeding 1 billion in revenue.

**Average Box Office by Genre**: Animation films have the highest average revenue, likely driven by family friendly content that appeals to larger audiences. Action films also perform well on average, while drama and biography films tend to have lower average revenue (like Oppenheimer).


## 2. Pie Charts


```python
genre_counts = df['Genre'].value_counts()
plt.figure(figsize=(10, 8))
plt.pie(genre_counts.values, labels=genre_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Distribution of Movies by Genre')
plt.axis('equal')
plt.tight_layout()
plt.show()
plt.close()
```

```python
year_counts = df['Year'].value_counts().sort_index()
plt.figure(figsize=(10, 8))
plt.pie(year_counts.values, labels=year_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Distribution of Movies by Release Year')
plt.axis('equal')
plt.tight_layout()
plt.show()
plt.close()
```

### Interpretation of Pie Charts

**Distribution by Genre**: Action films make up the largest portion of the dataset, reflecting the dominance of action movies in. Animation and Sci-Fi films each represent significant parts, while Drama, Biography, Comedy, and Musical genres aren't that common.

**Distribution by Year**: The data shows a larger amount of movies from 2022 and 2023, with 2022 having the largest amount.


## 3. Bubble Charts / Scatter Plots


```python
plt.figure(figsize=(12, 8))
scatter = plt.scatter(df['Rating'], df['Box Office (M$)'], 
                     s=df['Duration (min)']*3, alpha=0.6, 
                     c=df['Year'], cmap='viridis')
plt.xlabel('Rating')
plt.ylabel('Box Office Revenue (Million $)')
plt.title('Rating vs Box Office Revenue (Bubble size = Duration in minutes)')
plt.colorbar(scatter, label='Year')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
plt.close()
```

```python
genres = df['Genre'].unique()
colors = plt.cm.Set3(np.linspace(0, 1, len(genres)))
genre_color_map = dict(zip(genres, colors))

plt.figure(figsize=(12, 8))
for genre in genres:
    genre_data = df[df['Genre'] == genre]
    plt.scatter(genre_data['Duration (min)'], genre_data['Box Office (M$)'], 
               label=genre, s=100, alpha=0.7)

plt.xlabel('Duration (minutes)')
plt.ylabel('Box Office Revenue (Million $)')
plt.title('Duration vs Box Office Revenue by Genre')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
plt.close()
```

### Interpretation of Bubble Charts / Scatter Plots

**Rating vs Box Office (with Duration)**: There is a weak positive correlation between rating and revenue performance. Higher-rated films (above 8.0) tend to perform well, but some lower rated films (like Barbie at 6.8) achieved good revenue success ($1.4 billion). The bubble sizes reveal that longer films don't necessarily correlate with higher revenue, though some of the highest earning films (Avatar, Oppenheimer) are longer.

**Duration vs Box Office by Genre**: Most films cluster between 100-180 minutes. Action and Sci-Fi films show wide variation in both duration and revenue performance. Animation films tend to be shorter (under 100 minutes) but can achieve high returns. There's no clear linear relationship between duration and success across genres.


## 4. Box Plots


```python
df.boxplot(column='Rating', by='Genre', figsize=(12, 6))
plt.title('Rating Distribution by Genre')
plt.suptitle('')
plt.xlabel('Genre')
plt.ylabel('Rating')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
plt.close('all')
```

```python
df.boxplot(column='Box Office (M$)', by='Genre', figsize=(12, 6))
plt.title('Box Office Revenue Distribution by Genre')
plt.suptitle('')
plt.xlabel('Genre')
plt.ylabel('Box Office Revenue (Million $)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
plt.close('all')
```

```python
df.boxplot(column='Duration (min)', by='Genre', figsize=(12, 6))
plt.title('Duration Distribution by Genre')
plt.suptitle('')
plt.xlabel('Genre')
plt.ylabel('Duration (minutes)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
plt.close('all')
```

### Interpretation of Box Plots

**Rating Distribution by Genre**: Biography and Sci-Fi genres show the highest median ratings (around 8.0-8.3), indicating these genres tend to receive better critique. Action films have a wide range of ratings (from 5.6 to 8.2), showing variation in ratings. Animation films are around 7.0, suggesting consistent but not amazing ratings.

**Box Office Distribution by Genre**: Animation and Action genres show the highest median revenue performance. Drama and Biography films have lower median revenue but include some exceptions. The wide spread in Action films indicates that not all action movies are guaranteed success.

**Duration Distribution by Genre**: Animation films consistently have shorter durations (median around 90-100 minutes), while Biography and Drama films tend to be longer (median around 180+ minutes). Action and Sci-Fi films show difference in duration, usually from 120 to 180 minutes.
