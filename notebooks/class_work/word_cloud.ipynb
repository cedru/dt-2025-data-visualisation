# Text Processing and Word Cloud

Two small tasks in one notebook:
- Task 1: fetch text (from a URL with a safe fallback), do quick cleaning, and inspect the most frequent words.
- Task 2: create a basic word cloud from that processed text and comment on what it reveals.


```python
import re
from collections import Counter

import matplotlib.pyplot as plt
import requests
from wordcloud import STOPWORDS, WordCloud
```

```python
# Task 1 – Fetch text from a URL (with a fallback) and preview it
url = "https://www.gutenberg.org/cache/epub/11/pg11.txt"  # Alice in Wonderland
fallback_text = (
    "Alice was beginning to get very tired of sitting by her sister on the bank, "
    "and of having nothing to do. Once or twice she had peeped into the book her sister "
    "was reading, but it had no pictures or conversations in it, 'and what is the use of a book,' "
    "thought Alice 'without pictures or conversation?'"
)

try:
    response = requests.get(url, timeout=10)
    response.raise_for_status()
    raw_text = response.text
    source = f"Downloaded from {url}"
except Exception as exc:  # pragma: no cover - fallback for offline use
    raw_text = fallback_text
    source = f"Fallback text used because: {exc}"

print(source)
print("\nPreview (first 400 chars):\n" + raw_text[:400])
print(f"\nTotal characters: {len(raw_text):,}")
```

```python
# Basic cleaning: lowercase, drop non-letters, remove stopwords, and count
stopwords = set(STOPWORDS)
cleaned_text = re.sub(r"[^a-zA-Z\s]", " ", raw_text).lower()
words = [w for w in cleaned_text.split() if w not in stopwords and len(w) > 2]

word_counts = Counter(words)
print(f"Total words after cleaning: {len(words):,}")
print("Top 10 words:")
for word, count in word_counts.most_common(10):
    print(f"{word:>12}: {count}")
```

## Task 2 – Word Cloud
Generate a quick word cloud from the cleaned text.


```python
wc = WordCloud(
    width=900,
    height=450,
    background_color="white",
    max_words=120,
    stopwords=stopwords,
    colormap="plasma",
).generate(" ".join(words))

plt.figure(figsize=(12, 6))
plt.imshow(wc, interpolation="bilinear")
plt.axis("off")
plt.title("Word Cloud from Alice in Wonderland (or fallback text)")
plt.show()
```

### Interpretation
The largest words in the cloud point to recurring themes and characters. Expect names like "alice" and scene-related words (e.g., "queen", "king", "time") to dominate, showing who and what appears most often. Smaller, scattered words hint at secondary motifs and actions, giving a quick visual sense of what the text focuses on without reading every line.
